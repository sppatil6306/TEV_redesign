```python
import os, re
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from scipy import stats

import biotite.structure as struc
import biotite.structure.io.pdb as pdb

from psa import load, elastic, spatial
```

```python
def get_ca_coords_and_resids(pps, chain_idx=0):
    """Extract CA coordinates and residue IDs from protein."""
    coords, resids = [], []
    for pp in pps[chain_idx]:
        for res in pp:
            try:
                ca = res['CA'].get_vector().get_array()
                coords.append(ca)
                resids.append(res.get_id()[1])
            except KeyError:
                pass
    return np.array(coords), np.array(resids, dtype=int)

def kabsch_align(ref_coords, mobile_coords):
    """Align mobile to reference using Kabsch algorithm (SVD)."""
    ref_centered = ref_coords - ref_coords.mean(axis=0)
    mobile_centered = mobile_coords - mobile_coords.mean(axis=0)
    H = mobile_centered.T @ ref_centered
    U, _, Vt = np.linalg.svd(H)
    R = Vt.T @ U.T
    if np.linalg.det(R) < 0:
        Vt[-1, :] *= -1
        R = Vt.T @ U.T
    mobile_aligned = (mobile_centered @ R.T) + ref_coords.mean(axis=0)
    rmsd = np.sqrt(np.mean((mobile_aligned - ref_coords) ** 2))
    return mobile_aligned, rmsd

def compute_psa_strain(ref_coords, des_coords, elastic_cutoff=10.0):
    """Compute PSA strain using elastic network model."""
    if len(ref_coords) < 3:
        return np.nan, None, None
    try:
        weights = elastic.compute_weights_fast(
            [ref_coords, des_coords],
            "intersect",
            [elastic_cutoff]
        )
        F = elastic.deformation_gradient_fast(weights, ref_coords, des_coords)
        gam_l, gam_n = elastic.lagrange_strain(F)
        stretches, stretch_axis = elastic.principal_stretches_from_g(gam_n)
        strain_score = float(np.linalg.norm(gam_n))
        return strain_score, gam_n, stretches
    except Exception as e:
        return np.nan, None, None

def match_residues(ref_coords, ref_resids, des_coords, des_resids, target_resids):
    """Match residues between reference and design."""
    ref_matched, des_matched = [], []
    for rid in target_resids:
        ref_idx = np.where(ref_resids == rid)[0]
        des_idx = np.where(des_resids == rid)[0]
        if len(ref_idx) == 1 and len(des_idx) == 1:
            ref_matched.append(ref_coords[ref_idx[0]])
            des_matched.append(des_coords[des_idx[0]])
    if not ref_matched:
        return np.array([]), np.array([]), 0
    return np.array(ref_matched), np.array(des_matched), len(ref_matched)

```

```python
PDB_DIR = '/content/'

# Load reference
ref_pps = load.single_structure(name='1LVM', path=PDB_DIR, file_type='pdb')
ref_ca_coords, ref_resids = get_ca_coords_and_resids(ref_pps)

print(f"✓ Reference loaded: {len(ref_ca_coords)} CA atoms")
print(f"  Residue range: {ref_resids.min()}-{ref_resids.max()}")
```

```python
CATALYTIC_RES = [46, 81, 151]
SHELL_CUTOFF = 10.0

# Find catalytic atoms
cat_mask = np.isin(ref_resids, CATALYTIC_RES)
cat_coords = ref_ca_coords[cat_mask]

# Create shell
shell_mask = np.zeros(len(ref_ca_coords), dtype=bool)
for cat_coord in cat_coords:
    distances = np.linalg.norm(ref_ca_coords - cat_coord, axis=1)
    shell_mask |= (distances <= SHELL_CUTOFF)

shell_indices = np.where(shell_mask)[0]
shell_resids = ref_resids[shell_indices]
ref_shell_coords = ref_ca_coords[shell_indices]

print(f"✓ Catalytic residues: {len(cat_coords)} atoms")
print(f"✓ Active site shell: {len(shell_resids)} atoms")
```

```python
# Get all PDB files in uploads
all_files = os.listdir(PDB_DIR)
pdb_files = [f for f in all_files if f.endswith('.pdb') and not f.startswith('1LVM')]

print(f"✓ Found {len(pdb_files)} design PDB files")
print(f"  Sample: {pdb_files[:3]}")
```

```python
results = []
failed_designs = []

ELASTIC_CUTOFF = 10.0

print(f"\nProcessing {len(pdb_files)} designs...")
print("=" * 70)

for i, fname in enumerate(pdb_files, 1):
    # Progress indicator
    if i % 20 == 1 or i == len(pdb_files):
        print(f"[{i:3d}/{len(pdb_files)}] Processing...")

    try:
        # Extract design ID from filename
        match = re.match(r'(hyperTEV\d+)', fname)
        if not match:
            continue
        design_id = match.group(1)

        #Load design structure
        des_pps = load.single_structure(
            name=fname.replace('.pdb', ''),
            path=PDB_DIR,
            file_type='pdb'
        )
        des_ca_coords, des_resids = get_ca_coords_and_resids(des_pps)

        if len(des_ca_coords) < 10:
            failed_designs.append((design_id, f"Only {len(des_ca_coords)} atoms"))
            continue

        # Match active site residues
        ref_shell, des_shell, n_matched = match_residues(
            ref_ca_coords, ref_resids,
            des_ca_coords, des_resids,
            shell_resids
        )

        if n_matched < 8:
            failed_designs.append((design_id, f"Only {n_matched} residues matched"))
            continue

        #Align
        des_shell_aligned, rmsd = kabsch_align(ref_shell, des_shell)

        #Compute strain
        strain_score, gam_n, stretches = compute_psa_strain(
            ref_shell,
            des_shell_aligned,
            elastic_cutoff=ELASTIC_CUTOFF
        )

        if np.isnan(strain_score):
            failed_designs.append((design_id, "PSA computation failed"))
            continue

        results.append({
            'design_id': design_id,
            'filename': fname,
            'strain_score': strain_score,
            'rmsd_alignment': rmsd,
            'n_matched_residues': n_matched
        })

    except Exception as ex:
        match = re.match(r'(hyperTEV\d+)', fname)
        if match:
            design_id = match.group(1)
            failed_designs.append((design_id, str(ex)[:50]))

# Convert to DataFrame
results_df = pd.DataFrame(results)
if len(failed_designs) > 0 and len(failed_designs) <= 10:
    print(f"\nFailures:")
    for design_id, reason in failed_designs:
        print(f"  {design_id}: {reason}")
elif len(failed_designs) > 10:
    print(f"\nFirst 10 failures:")
    for design_id, reason in failed_designs[:10]:
        print(f"  {design_id}: {reason}")
```

```python

activity_col = 'greater than 5% of WT activity'

# Convert to boolean
df_activity['active'] = df_activity[activity_col].astype(bool)

print("Activity breakdown:")
print(df_activity['active'].value_counts())
print(f"\n  Active (True): {(df_activity['active'] == True).sum()}")
print(f"  Inactive (False): {(df_activity['active'] == False).sum()}")
```

```python
merged_df = results_df.merge(
    df_activity[['design_id', 'active']],
    on='design_id',
    how='left'
)

print(f"Merged results:")
print(f"  Total designs: {len(merged_df)}")
print(f"  With activity labels: {merged_df['active'].notna().sum()}")
print(f"  Missing labels: {merged_df['active'].isna().sum()}")

print(f"\nFirst few rows:")
print(merged_df.head(10))
```

```python
# Remove rows with missing activity labels
df_clean = merged_df.dropna(subset=['active']).copy()

# Separate by activity
active_strains = df_clean[df_clean['active'] == True]['strain_score'].values
inactive_strains = df_clean[df_clean['active'] == False]['strain_score'].values

print(f"Active designs (>5% WT activity):")
print(f"  Count: {len(active_strains)}")
print(f"  Mean strain: {active_strains.mean():.6f}")
print(f"  Median strain: {np.median(active_strains):.6f}")
print(f"  Std dev: {active_strains.std():.6f}")
print(f"  Range: {active_strains.min():.6f} - {active_strains.max():.6f}")

print(f"\nInactive designs (≤5% WT activity):")
print(f"  Count: {len(inactive_strains)}")
print(f"  Mean strain: {inactive_strains.mean():.6f}")
print(f"  Median strain: {np.median(inactive_strains):.6f}")
print(f"  Std dev: {inactive_strains.std():.6f}")
print(f"  Range: {inactive_strains.min():.6f} - {inactive_strains.max():.6f}")

print(f"\nDifference:")
print(f"  Mean difference: {inactive_strains.mean() - active_strains.mean():.6f}")
print(f"  Median difference: {np.median(inactive_strains) - np.median(active_strains):.6f}")
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
from scipy import stats

# Perform Mann-Whitney U test
# This assumes 'active_strains' and 'inactive_strains' are defined from previous cells
stat, p_value = stats.mannwhitneyu(active_strains, inactive_strains, alternative='two-sided')

fig, ax = plt.subplots(figsize=(12, 7))

# Separate data
active_data = df_clean[df_clean['active'] == True]
inactive_data = df_clean[df_clean['active'] == False]

# Add jitter
jitter_strength = 0.04
active_jitter = 1 + np.random.normal(0, jitter_strength, len(active_data))
inactive_jitter = 0 + np.random.normal(0, jitter_strength, len(inactive_data))

# Scatter plots
ax.scatter(inactive_jitter, inactive_data['strain_score'],
          label=f'Inactive (n={len(inactive_data)})',
          color='#e74c3c', s=120, alpha=0.6, edgecolors='black', linewidth=0.7)

ax.scatter(active_jitter, active_data['strain_score'],
          label=f'Active (n={len(active_data)})',
          color='#2ecc71', s=120, alpha=0.6, edgecolors='black', linewidth=0.7)

# Add horizontal lines for medians
median_inactive = np.median(inactive_strains)
median_active = np.median(active_strains)

ax.hlines(median_inactive, -0.3, 0.3, colors='#c0392b', linestyles='-', linewidth=3, label=f'Inactive median: {median_inactive:.4f}')
ax.hlines(median_active, 0.7, 1.3, colors='#27ae60', linestyles='-', linewidth=3, label=f'Active median: {median_active:.4f}')

# Add mean lines (optional)
ax.hlines(np.mean(inactive_strains), -0.3, 0.3, colors='#c0392b', linestyles=':', linewidth=2, alpha=0.5)
ax.hlines(np.mean(active_strains), 0.7, 1.3, colors='#27ae60', linestyles=':', linewidth=2, alpha=0.5)

# Formatting
ax.set_xlim(-0.5, 1.5)
ax.set_xticks([0, 1])
ax.set_xticklabels(['Inactive\n(≤5% WT Activity)', 'Active\n(>5% WT Activity)'],
                   fontsize=13, fontweight='bold')

ax.set_ylabel('PSA Strain Score', fontsize=13, fontweight='bold')
ax.set_title('Active vs Inactive Designs: PSA Strain Comparison', fontsize=15, fontweight='bold')
ax.grid(axis='y', alpha=0.3, linestyle='--')

# Statistics box
stats_text = f'Mann-Whitney U Test\np-value = {p_value:.6f}'
if p_value < 0.05:
    stats_text += '\n✓ SIGNIFICANT'
    box_color = '#2ecc71'
else:
    stats_text += '\n✗ Not Significant'
    box_color = '#e74c3c'

ax.text(0.5, ax.get_ylim()[1] * 0.92, stats_text,
        ha='center', fontsize=11, fontweight='bold',
        bbox=dict(boxstyle='round', facecolor=box_color, alpha=0.3, edgecolor='black', linewidth=2))

ax.legend(fontsize=11, loc='upper left', framealpha=0.95)
plt.tight_layout()
plt.savefig('strain_comparison_scatter_styled.png', dpi=300, bbox_inches='tight')
print("✓ Saved: strain_comparison_scatter_styled.png")
plt.show()
```
