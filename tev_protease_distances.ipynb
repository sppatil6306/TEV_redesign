```python
import pandas as pd

df = pd.read_csv("hyperTEV_rates (1).csv")  # replace with your actual filename
df.head()
```

```python
import glob
import biotite.structure.io.pdb as pdb
import biotite.structure as struc  # needed for AtomArrayStack

# Normalize CSV design names
df["design_norm"] = df["design"].str.strip().str.lower()

structures = []
activities = []
filenames = []

for filepath in glob.glob("*.pdb"):
    filename = filepath.split("/")[-1]

    # Take everything before the first underscore
    design_name_file = filename.replace(".pdb", "").split("_")[0].strip().lower()

    # Match CSV row
    row = df[df["design_norm"] == design_name_file]
    if len(row) == 0:
        print(f"Skipping {filename} — no CSV match")
        continue

    activity = row["greater than 5% of WT activity"].values[0]

    # Load PDB
    pdb_file = pdb.PDBFile.read(filepath)
    structure_stack = pdb_file.get_structure(model=None)  # load all models

    # Use only first model if AtomArrayStack
    if isinstance(structure_stack, struc.AtomArrayStack):
        structure = structure_stack[0]
    else:
        structure = structure_stack

    structures.append(structure)
    activities.append(str(activity).upper() == "TRUE")
    filenames.append(filename)

print(f"Matched {len(structures)} structures to CSV")
```

```python
import numpy as np
import pandas as pd # Added pandas import

def get_atom_coord(structure, res_id, atom_name):
    """
    Returns the coordinate of a given atom in a residue.
    """
    mask = (structure.res_id == res_id) & (structure.atom_name == atom_name)
    coords = structure.coord[mask]
    if len(coords) == 0:
        raise ValueError(f"Atom {atom_name} in residue {res_id} not found")
    return coords[0]

# Initialize lists to store all distances
cys_his_distances_list = []
his_asp_distances_list = []
cys_asp_distances_list = []

# List to store all data for the DataFrame
all_data = []

for i, s in enumerate(structures):
    # Retrieve filename and activity for the current structure
    current_filename = filenames[i]
    current_activity = activities[i]

    row_data = {
        "filename": current_filename,
        "activity": current_activity
    }

    try:
        cys_s = get_atom_coord(s, 151, "SG")  # Cys151 sulfur
        his_n = get_atom_coord(s, 46, "NE2")  # His46 nitrogen
        # Assuming Asp150 as the catalytic Asp, using CG atom (check PDB for exact residue ID and atom)
        asp_c = get_atom_coord(s, 150, "CG")  # Asp150 gamma carbon

        cys_his_dist = np.linalg.norm(cys_s - his_n)
        his_asp_dist = np.linalg.norm(his_n - asp_c)
        cys_asp_dist = np.linalg.norm(cys_s - asp_c)

        cys_his_distances_list.append(cys_his_dist)
        his_asp_distances_list.append(his_asp_dist)
        cys_asp_distances_list.append(cys_asp_dist)

        row_data["cys_his_dist"] = cys_his_dist
        row_data["his_asp_dist"] = his_asp_dist
        row_data["cys_asp_dist"] = cys_asp_dist

    except ValueError as e:
        print(f"Skipping {current_filename} due to missing atom: {e}")
        cys_his_distances_list.append(np.nan)
        his_asp_distances_list.append(np.nan)
        cys_asp_distances_list.append(np.nan)

        row_data["cys_his_dist"] = np.nan
        row_data["his_asp_dist"] = np.nan
        row_data["cys_asp_dist"] = np.nan

    all_data.append(row_data)

# Convert the Cys-His distances list to a numpy array for compatibility with previous code
distances = np.array(cys_his_distances_list)

# Convert other distance lists to numpy arrays (optional, but good for consistency)
his_asp_distances = np.array(his_asp_distances_list)
cys_asp_distances = np.array(cys_asp_distances_list)

# Create a DataFrame with all distances and associated metadata
all_distances_df = pd.DataFrame(all_data)

print("Cys-His distances array 'distances' updated. New DataFrame 'all_distances_df' created:")
display(all_distances_df.head())
```

```python
import numpy as np

filenames = np.array(filenames)
activities = np.array(activities)

inactive_designs = filenames[activities == False]

print(f"Number of inactive designs: {len(inactive_designs)}")
inactive_designs
```

```python
inactive_mask = (activities == False)
inactive_dist = distances[inactive_mask]

# Remove NaNs
inactive_dist = inactive_dist[~np.isnan(inactive_dist)]

over_4 = inactive_dist > 4.0

print(f"Inactive designs total: {len(inactive_dist)}")
print(f"Inactive > 4.0 Å: {np.sum(over_4)}")
print(f"Fraction drifting > 4.0 Å: {np.mean(over_4):.2f}")
```
